# 2026 NFL Draft Big Board

Combined round predictions from all position models (CB, S, LB, ED, DT, OT, IOL, RB). Players are ranked by **predicted round** (1 = first round, 8 = undrafted).

**To refresh predictions:** Run each position's round regression notebook so they save updated CSVs to `data/processed/`:
- `CB/cb_round_regression.ipynb`
- `S/s_round_regression.ipynb`
- `LB/lb_round_regression.ipynb`
- `Edges/edges_round_regression.ipynb`
- `DT/dt_round_regression.ipynb`
- `OT/ot_round_regression.ipynb`
- `IOL/iol_round_regression.ipynb`
- `RB/rb_round_regression.ipynb`

```python
import pandas as pd
import os

# Show full dataframe (no row/column truncation)
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)

DATA = 'data/processed'
pred_files = [
    ('cb_2026_predictions.csv', 'CB'),
    ('s_2026_predictions.csv', 'S'),
    ('lb_2026_predictions.csv', 'LB'),
    ('edges_2026_predictions.csv', 'ED'),
    ('dt_2026_predictions.csv', 'DT'),
    ('ot_2026_predictions.csv', 'OT'),
    ('iol_2026_predictions.csv', 'IOL'),
    ('rb_2026_predictions.csv', 'RB'),
]

dfs = []
for filename, pos in pred_files:
    path = os.path.join(DATA, filename)
    if os.path.exists(path):
        df = pd.read_csv(path)
        if 'Pos' not in df.columns:
            df['Pos'] = pos
        dfs.append(df)
    else:
        print(f"Missing {path} — run the {pos} round regression notebook to generate it.")

if not dfs:
    raise SystemExit("No prediction files found. Run each position's round_regression notebook first.")

big = pd.concat(dfs, ignore_index=True)
big = big.sort_values(['predicted_round', 'Player']).reset_index(drop=True)
big.insert(0, 'rank', big.index + 1)

# Reorder columns: rank, Player, Pos, School, predicted_round
cols = [c for c in ['rank', 'Player', 'Pos', 'School', 'predicted_round'] if c in big.columns]
big_board = big[cols].copy()
big_board['predicted_round'] = big_board['predicted_round'].astype(float)

print(f"Big board: {len(big_board)} players from {len(dfs)} position groups")
print(f"Predicted rounds: {big_board['predicted_round'].min():.0f} – {big_board['predicted_round'].max():.0f} (8 = undrafted)")
display(big_board)
```
